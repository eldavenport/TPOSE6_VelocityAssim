```python
# get diag from previous window output 
import xarray as xr
from xmitgcm import open_mdsdataset
```

```python
month_name = 'jul012013'
tpose_month = 'mar2013' # end of march window is 1 day of july 
folder_name = 'model_setup/forecast_IC/' + month_name + '/'

data_dir = '/data/SO3/edavenport/tpose6/' + tpose_month + '/run_iter16/'
grid_dir = '/data/SO6/TPOSE_diags/tpose6/grid_6/'

# grab the last iteration
iter = 8784 

tpose_ds = open_mdsdataset(data_dir=data_dir,grid_dir=grid_dir,iters=iter,prefix='diag_state')
```

```python
tpose_ds = tpose_ds.drop_vars(['DRHODR','WVEL'])
```

```python
tpose_ds.data_vars
```

```python
for v in tpose_ds.data_vars:
    arr = tpose_ds[v].values
    # print(folder_name + f'{v}_assim_jan2013.data')
    arr.astype('>f4').tofile(folder_name + f'{v}_tpose_Vel_'+ month_name +'.data')
```
